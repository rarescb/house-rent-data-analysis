```python
#Pregatirea datelor pentru model

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn import metrics

# 1. Separarea în Predictor (X) și Țintă (y)
X_simplu = df[['Size']]
y_simplu = df['Rent']

# 2. Segmentarea datelor în eșantioane de antrenare și testare
X_train_s, X_test_s, y_train_s, y_test_s = train_test_split(X_simplu, y_simplu, test_size=0.3, random_state=101)

X_simplu.shape
y_simplu.shape
```

```python
#Modelul 1 Regresie Liniara Simpla

lm_simplu = LinearRegression()
lm_simplu.fit(X_train_s, y_train_s)


#Afișarea coeficienților (structura internă a modelului)
print("=== PARAMETRII STRUCTURALI (MODEL SIMPLU) ===")
print(f"Intercept (Beta_0): {lm_simplu.intercept_:.2f}")
print(f"Coeficient de pantă (Beta_1 pentru Income): {lm_simplu.coef_[0]:.4f}")

# Generarea predicțiilor și metricile numerice de evaluare a calității
pred_simplu = lm_simplu.predict(X_test_s)
mae_simplu = metrics.mean_absolute_error(y_test_s, pred_simplu)
rmse_simplu = np.sqrt(metrics.mean_squared_error(y_test_s, pred_simplu))
r2_simplu = metrics.r2_score(y_test_s, pred_simplu)

print("\n=== METRICI DE CALITATE NUMERICĂ ===")
print(f"MAE (Eroarea Absolută Medie): ${mae_simplu:,.2f}")
print(f"RMSE (Rădăcina Erorii Pătratice Medii): ${rmse_simplu:,.2f}")
print(f"Scorul R² (Coeficientul de Determinare): {r2_simplu:.4f}")

```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error
from sklearn.model_selection import learning_curve

# ==========================================
# 1. CITIREA SETULUI DE DATE REAL
# ==========================================
df_real = pd.read_csv('house.csv')

# 2. APLICAREA PROTOCOLULUI DE CURĂȚARE (FILTRAREA IQR DIN PROIECTUL 2)
# ==========================================
df_clean_real = df_real.copy()
numeric_cols = ['BHK','Rent','Size','Bathroom']

# Eliminăm zgomotul și valorile aberante (Outliers) din datele reale
for col in numeric_cols:
    Q1 = df_clean_real[col].quantile(0.25)
    Q3 = df_clean_real[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    df_clean_real = df_clean_real[(df_clean_real[col] >= lower_limit) & (df_clean_real[col] <= upper_limit)]
# ==========================================
# 3. SEPARAREA MATRICELOR 
# ==========================================
# Izolăm exclusiv cei 5 predictori numerici curați
X_real_raw = df_clean_real[['BHK','Size','Bathroom']]
y_real = df_clean_real['Rent']

# Aplicarea transformării polinomiale de gradul 3 pentru a genera complexitate structurală (55 de predictori)
poly_transformer = PolynomialFeatures(degree=3, include_bias=False)
X_real_poly = poly_transformer.fit_transform(X_real_raw)


# Segmentarea asimetrică 
X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(X_real_poly, y_real, train_size=70, random_state=42)

# Standardizarea 
scaler_real = StandardScaler()
X_train_r_scaled = scaler_real.fit_transform(X_train_r)
X_test_r_scaled = scaler_real.transform(X_test_r)

print("\n=== STATISTICI FINALE PREGĂTIRE DATE REALE ===")
print(f"Număr de predictori rezultați după extinderea polinomială: {X_real_poly.shape[1]}")
print(f"Dimensiune eșantion antrenare (X_train pentru fit):        {X_train_r_scaled.shape}")
print(f"Dimensiune eșantion validare (X_test pentru predict):      {X_test_r_scaled.shape}")
```

```python
# ==========================================
# CURBA DE ÎNVĂȚARE PENTRU SCORUL R²
# ==========================================
from sklearn.model_selection import learning_curve
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# 1. Calcularea curbei de învățare folosind scoring='r2'
train_sizes_o, train_scores_o, test_scores_o = learning_curve(
    LinearRegression(), X_train_r_scaled, y_train_r, cv=3, scoring='r2',
    train_sizes=np.linspace(0.2, 1.0, 8), random_state=101
)

# 2. Plotarea Curbei de Învățare R²
plt.figure(figsize=(9, 4.5))
plt.plot(train_sizes_o, train_scores_o.mean(axis=1), 'o-', color='#800080', lw=2, label='Train R² (Overfitting)')
plt.plot(train_sizes_o, test_scores_o.mean(axis=1), 'o-', color='#FF00FF', lw=2, label='Test R² (Colaps pe minus)')

plt.title('Curba de Învățare OLS — Scor R² (Fără Regularizare)', fontsize=12, fontweight='bold')
plt.xlabel('Dimensiune Set Antrenare')
plt.ylabel('Scor R²')
plt.ylim(-6, 1.1)  
plt.legend()
plt.grid(True, ls='--', alpha=0.5)
plt.show()
```

```python
# ==========================================
# CURBA DE ÎNVĂȚARE PENTRU EROAREA RMSE
# ==========================================
from sklearn.model_selection import learning_curve
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# 1. Calcularea curbei de învățare folosind RMSE (negat implicit de sklearn)
train_sizes_rmse, train_scores_rmse, test_scores_rmse = learning_curve(
    LinearRegression(), X_train_r_scaled, y_train_r, cv=3, scoring='neg_root_mean_squared_error',
    train_sizes=np.linspace(0.2, 1.0, 8), random_state=101
)

# 2. Convertirea scorurilor negative în valori RMSE pozitive reale
train_rmse_clean = -train_scores_rmse.mean(axis=1)
test_rmse_clean = -test_scores_rmse.mean(axis=1)

# 3. Plotarea Curbei de Învățare RMSE
plt.figure(figsize=(9, 4.5))
plt.plot(train_sizes_rmse, train_rmse_clean, 'o-', color='#708D23', lw=2, label='Train RMSE (Eroare mică)')
plt.plot(train_sizes_rmse, test_rmse_clean, 'o-', color='#BEF574', lw=2, label='Test RMSE (Explozie Eroare)')

plt.title('Curba de Învățare OLS — Eroare RMSE (Fără Regularizare)', fontsize=12, fontweight='bold')
plt.xlabel('Dimensiune Set Antrenare')
plt.ylabel('Eroare RMSE ($)')
plt.legend()
plt.grid(True, ls='--', alpha=0.5)
plt.show()
```

```python
from sklearn.linear_model import Ridge, Lasso
from sklearn.model_selection import GridSearchCV, KFold
from sklearn.metrics import r2_score, mean_squared_error

# 1. Definirea grilelor de căutare 
param_grid_alpha = {'alpha': np.logspace(-2, 4, 40)}
cv_real = KFold(n_splits=3, shuffle=True, random_state=101)

# 2. Instanțierea și rularea optimizării pentru toate tipurile de regularizare
ridge_gs_real = GridSearchCV(Ridge(), param_grid_alpha, cv=cv_real, scoring='r2', n_jobs=-1)
lasso_gs_real = GridSearchCV(Lasso(max_iter=20000), param_grid_alpha, cv=cv_real, scoring='r2', n_jobs=-1)

ridge_gs_real.fit(X_train_r_scaled, y_train_r)
lasso_gs_real.fit(X_train_r_scaled, y_train_r)

# 3. Extragerea celor mai buni estimatori gata acordați fin
best_ridge_r = ridge_gs_real.best_estimator_
best_lasso_r = lasso_gs_real.best_estimator_

print("=== HIPERPARAMETRII OPTIMI IDENTIFICAȚI ===")
print(f"Top Ridge -> Alpha: {ridge_gs_real.best_params_['alpha']:.2f}")
print(f"Top Lasso -> Alpha: {lasso_gs_real.best_params_['alpha']:.2f}")

```

```python
# ==========================================
# RIDGE (L2) OPTIMIZAT
# ==========================================
from sklearn.model_selection import learning_curve
import numpy as np
import matplotlib.pyplot as plt

# Folosește estimatorul optim extras automat de GridSearchCV la celula anterioară
sizes_ridge, train_scores_ridge, test_scores_ridge = learning_curve(
    best_ridge_r, X_train_r_scaled, y_train_r, cv=3, scoring='r2',
    train_sizes=np.linspace(0.1, 1.0, 8), random_state=101
)

plt.figure(figsize=(9, 4.5))
plt.plot(sizes_ridge, train_scores_ridge.mean(axis=1), 'o-', color='#708D23', lw=2, label='Train R²')
plt.plot(sizes_ridge, test_scores_ridge.mean(axis=1), 'o-', color='#01796F', lw=2, label='Test R²')

plt.title('Curba de Învățare: Ridge (L2) Optimizat — Stabilizare', fontsize=12, fontweight='bold')
plt.xlabel('Dimensiune Set Antrenare')
plt.ylabel('Scor R²')
plt.ylim(0, 1.1) 
plt.legend(loc='lower right')
plt.grid(True, ls='--', alpha=0.5)
plt.show()
```

```python
# ==========================================
#  LASSO (L1) OPTIMIZAT
# ==========================================
from sklearn.model_selection import learning_curve
import numpy as np
import matplotlib.pyplot as plt

# Folosește estimatorul optim lasso extras automat de GridSearchCV la celula anterioară
sizes_lasso, train_scores_lasso, test_scores_lasso = learning_curve(
    best_lasso_r, X_train_r_scaled, y_train_r, cv=3, scoring='r2',
    train_sizes=np.linspace(0.1, 1.0, 8), random_state=101
)

plt.figure(figsize=(9, 4.5))
plt.plot(sizes_lasso, train_scores_lasso.mean(axis=1), 'o-', color='#2C3E50', lw=2, label='Train R²')
plt.plot(sizes_lasso, test_scores_lasso.mean(axis=1), 'o-', color='#27AE60', lw=2, label='Test R²')

plt.title('Curba de Învățare: Lasso (L1) Optimizat — Selectare Predictori', fontsize=12, fontweight='bold')
plt.xlabel('Dimensiune Set Antrenare')
plt.ylabel('Scor R²')
plt.ylim(0, 1.1)  
plt.legend(loc='lower right')
plt.grid(True, ls='--', alpha=0.5)
plt.show()
```

```python
# ==========================================
# TABEL COMPARATIV METRICI FINALE
# ==========================================
import pandas as pd
from sklearn.metrics import r2_score, mean_squared_error
import numpy as np

# 1. Calcularea predicțiilor pe setul de test pentru modelele optimizate
pred_ols_r = LinearRegression().fit(X_train_r_scaled, y_train_r).predict(X_test_r_scaled)
pred_ridge_r = best_ridge_r.predict(X_test_r_scaled)
pred_lasso_r = best_lasso_r.predict(X_test_r_scaled)

# 2. Centralizarea metricilor (R² și RMSE) pentru toate cele 3 scenarii
metrici_proiect = {
    'Model / Algoritm': [
        'Clasic OLS (Neregularizat)', 
        'Ridge (L2) Optimizat', 
        'Lasso (L1) Optimizat'
    ],
    'Scor R² (Test)': [
        r2_score(y_test_r, pred_ols_r), 
        r2_score(y_test_r, pred_ridge_r), 
        r2_score(y_test_r, pred_lasso_r)
    ],
    'Eroare RMSE ($)': [
        np.sqrt(mean_squared_error(y_test_r, pred_ols_r)),
        np.sqrt(mean_squared_error(y_test_r, pred_ridge_r)),
        np.sqrt(mean_squared_error(y_test_r, pred_lasso_r))
    ]
}

# 3. Crearea și formatarea vizuală a tabelului DataFrame
df_comparativ = pd.DataFrame(metrici_proiect)

# Formatăm numerele din tabel pentru a fi ușor de citit în raport
pd.options.display.float_format = '{:,.4f}'.format
print("=== REZUMATUL COMPARATIV AL PERFORMANȚELOR PE TEST ===")
display(df_comparativ)
```

```python
# ==========================================
# RIDGE (L2) OPTIMIZAT
# ==========================================
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score, mean_squared_error
import numpy as np

pred_ridge_r = best_ridge_r.predict(X_test_r_scaled)
r2_ridge = r2_score(y_test_r, pred_ridge_r)
rmse_ridge = np.sqrt(mean_squared_error(y_test_r, pred_ridge_r))

plt.figure(figsize=(8, 6))
plt.plot([y_test_r.min(), y_test_r.max()], [y_test_r.min(), y_test_r.max()], 'r--', lw=2, label='Predicție Ideală')
plt.scatter(y_test_r, pred_ridge_r, alpha=0.35, color='#CFA0E9', edgecolor='k', s=45, label='Date Test')

plt.title(f'Model Ridge (L2) Optimizat\nTest R² = {r2_ridge:.4f} | RMSE = ${rmse_ridge:,.0f}', fontsize=12, fontweight='bold')
plt.xlabel('Valori Reale (y_test)')
plt.ylabel('Predicții Model (y_pred)')
# Lăsăm axele libere; punctele se vor strânge strâns în jurul liniei ideale roșii
plt.grid(True, ls='--', alpha=0.4)
plt.legend()
plt.show()
```

```python
# ==========================================
# GRAFIC SCATTER 3: LASSO (L1) OPTIMIZAT
# ==========================================
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score, mean_squared_error
import numpy as np

pred_lasso_r = best_lasso_r.predict(X_test_r_scaled)
r2_lasso = r2_score(y_test_r, pred_lasso_r)
rmse_lasso = np.sqrt(mean_squared_error(y_test_r, pred_lasso_r))

plt.figure(figsize=(8, 6))
plt.plot([y_test_r.min(), y_test_r.max()], [y_test_r.min(), y_test_r.max()], 'r--', lw=2, label='Predicție Ideală')
plt.scatter(y_test_r, pred_lasso_r, alpha=0.35, color='#811453', edgecolor='k', s=45, label='Date Test')

plt.title(f'Model Lasso (L1) Optimizat\nTest R² = {r2_lasso:.4f} | RMSE = ${rmse_lasso:,.0f}', fontsize=12, fontweight='bold')
plt.xlabel('Valori Reale (y_test)')
plt.ylabel('Predicții Model (y_pred)')
plt.grid(True, ls='--', alpha=0.4)
plt.legend()
plt.show()
```

```python
# ==========================================
# GRAFIC 4: DISTRIBUȚIA ERORILOR REZIDUALE
# ==========================================
import matplotlib.pyplot as plt
import seaborn as sns

# Calculăm reziduurile pentru toate cele 3 modele
reziduuri_ols   = y_test_r - pred_ols_r
reziduuri_ridge = y_test_r - pred_ridge_r
reziduuri_lasso = y_test_r - pred_lasso_r

plt.figure(figsize=(9, 6))

# Randarea funcțiilor de densitate
sns.kdeplot(reziduuri_ols, fill=True, color='#048B9A', label='Reziduuri OLS (Erori Uriașe)', lw=2)
sns.kdeplot(reziduuri_ridge, fill=True, color='#6600FF', label='Reziduuri Ridge (Optimizat)', lw=2)
sns.kdeplot(reziduuri_lasso, fill=True, color='#6D071A', label='Reziduuri Lasso (Optimizat)', lw=2)

# Linie verticală de referință pentru eroare zero (Modelul Perfect)
plt.axvline(x=0, color='black', ls='--', alpha=0.7, label='Eroare Zero')

plt.title('Analiza Reziduurilor: Distribuția Erorilor de Predicție', fontsize=12, fontweight='bold')
plt.xlabel('Valoarea Erorii (Real - Prezis) [$]')
plt.ylabel('Densitate')

plt.grid(True, ls='--', alpha=0.4)
plt.legend()
plt.show()
```
